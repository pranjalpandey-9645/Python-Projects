```python
import numpy as np

# create a 1D Array
prices = np.array([10, 20, 30])

# The Numerical: Add 10% tax to every price at once
# In normal math: Price * 1.10
taxed_prices = prices * 1.10

print(f"Original: {prices}")
print(f"With Tax: {taxed_prices}")
```

```python
import numpy as np

#creating a 1D array
heights = np.array([60, 72, 68, 64])

# The numerical: convert all of them to centimeters
# hint: 1 inch = 2.54 cm
heights_in_centimeters = heights * 2.54

print(f"Original: {heights}")
print(f"In centimeters: {heights_in_centimeters}")
```
